```python
import math
```

```python
# Inputs
x1, x2 = 2.0, 3.0
y_true = 1   # actual value
```

```python
# Weights & Bias
w11, w12, b1 = 0.5, -0.2, 0.1
w21, w22, b2 = 0.8, 0.4, -0.1
w_out1, w_out2, b_out = 1.0, -1.5, 0.2
```

```python
# ReLU
def relu(z):
    return max(0, z)
```

```python
# Forward Pass
h1 = relu(x1*w11 + x2*w12 + b1)
h2 = relu(x1*w21 + x2*w22 + b2)

z_out = h1*w_out1 + h2*w_out2 + b_out
y_pred = 1 / (1 + math.exp(-z_out))   # Sigmoid
```

```python
# Loss (Binary Cross Entropy)
loss = - (y_true * math.log(y_pred) + (1 - y_true) * math.log(1 - y_pred))
```

```python
# Output
print("Prediction:", y_pred)
print("Loss:", loss)
```

```python
import matplotlib.pyplot as plt
# ---------------------------------------------------------
# 🔥 VISUALIZATION (Simple & Clean)
# ---------------------------------------------------------

values = [x1, x2, h1, h2, y_pred, loss]
labels = ["x1", "x2", "h1", "h2", "Pred", "Loss"]

plt.plot(values, marker='o')
plt.xticks(range(len(labels)), labels)
plt.title("Feed Forward + Loss Visualization")
plt.xlabel("Stages")
plt.ylabel("Values")
plt.grid()
plt.show()
```
