```python
import torch
import torch.nn as nn
```

```python
# Deep model with large weights
model = nn.Sequential(
    nn.Linear(1, 10),
    nn.ReLU(),
    nn.Linear(10, 10),
    nn.ReLU(),
    nn.Linear(10, 1)
)
```

```python
# Manually increase weights (to cause explosion)
for param in model.parameters():
    param.data *= 10
```

```python
# Input
x = torch.tensor([[1.0]])
y = torch.tensor([[2.0]])
```

```python
# Loss
loss_fn = nn.MSELoss()
```

```python
# Forward
output = model(x)
loss = loss_fn(output, y)
```

```python
# Backward
loss.backward()
```

```python
# Check gradients
for name, param in model.named_parameters():
    print(name, param.grad)
```
