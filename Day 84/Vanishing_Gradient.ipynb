```python
import torch
import torch.nn as nn
```

```python
# Simple Deep Model (with Sigmoid)
model = nn.Sequential(
    nn.Linear(1, 5),
    nn.Sigmoid(),
    nn.Linear(5, 5),
    nn.Sigmoid(),
    nn.Linear(5, 1)
)
```

```python
# Input & Output
x = torch.tensor([[1.0]])
y = torch.tensor([[2.0]])
```

```python
# Loss
loss_fn = nn.MSELoss()
```

```python
# Forward
output = model(x)
loss = loss_fn(output, y)
```

```python
# Backward
loss.backward()
```

```python
# Check gradients
for name, param in model.named_parameters():
    print(name, param.grad)
```
