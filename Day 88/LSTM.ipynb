```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# 1. Define LSTM + Linear
lstm = nn.LSTM(input_size=3, hidden_size=5, batch_first=True)
fc = nn.Linear(5, 1)
```

```python
# 2. Dummy Data
x = torch.randn(10, 4, 3)   # (batch=10, seq=4, features=3)
y = torch.randn(10, 1)      # target
```

```python

# 3. Loss & Optimizer
loss_fn = nn.MSELoss()
optimizer = optim.Adam(list(lstm.parameters()) + list(fc.parameters()), lr=0.01)
```

```python
# 4. Training Loop
for epoch in range(10):
    
    optimizer.zero_grad()
    
    # Forward
    out, (h, c) = lstm(x)
    
    # Take last time step
    out = out[:, -1, :]
    
    # Pass through Linear layer
    pred = fc(out)
    
    # Loss
    loss = loss_fn(pred, y)
    
    # Backward
    loss.backward()
    
    # Update
    optimizer.step()
    
    print("Epoch:", epoch, "Loss:", loss.item())
```
