```python
import torch
import torch.nn as nn
```

```python
# Model with Batch Normalization
model = nn.Sequential(
    nn.Linear(10, 20),
    nn.BatchNorm1d(20),  # BatchNorm layer
    nn.ReLU(),
    nn.Linear(20, 1)
)
```

```python
# Dummy data
x = torch.randn(5, 10)
```

```python
# Forward pass
output = model(x)

print(output)
```
