```python
import torch
import torch.nn as nn
```

```python
# Simple model with Dropout
model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Dropout(p=0.5),   # Dropout layer
    nn.Linear(20, 1)
)
```

```python
# Dummy data
x = torch.randn(5, 10)
```

```python
# Forward pass
output = model(x)

print(output)
```
