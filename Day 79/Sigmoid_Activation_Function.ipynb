```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sigmoid Function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# Generate input values
z = np.linspace(-10, 10, 200)
```

```python
# Apply Sigmoid
sigmoid_values = sigmoid(z)
```

```python
# Plot
plt.plot(z, sigmoid_values, linewidth=2)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input (z)")
plt.ylabel("Output")
plt.grid()
plt.show()
```
