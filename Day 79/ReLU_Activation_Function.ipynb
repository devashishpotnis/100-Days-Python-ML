```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ReLU Function
def relu(z):
    return np.maximum(0, z)
```

```python
# Generate input values
z = np.linspace(-10, 10, 200)
```

```python
# Apply ReLU
relu_values = relu(z)
```

```python
# Plot
plt.plot(z, relu_values, linewidth=2)
plt.title("ReLU Activation Function")
plt.xlabel("Input (z)")
plt.ylabel("Output")
plt.grid()
plt.show()
```
