```python
import pandas as pd
```

```python
# Dataset 1
customers = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "Name": ["Devashish", "Mangesh", "Amit"]
})

# Dataset 2
orders = pd.DataFrame({
    "CustomerID": [1, 2, 4],
    "OrderAmount": [500, 700, 300]
})
```

```python
#merge() (Most Important ⭐)
#Inner Merge
pd.merge(customers, orders, on="CustomerID")
```

```python
#Left Merge
pd.merge(customers, orders, on="CustomerID", how="left")
```

```python
#Right Merge
pd.merge(customers, orders, on="CustomerID", how="right")
```

```python
#Outer Merge
pd.merge(customers, orders, on="CustomerID", how="outer")
```

```python
#join()
customers.set_index("CustomerID").join(
    orders.set_index("CustomerID"),
    how="left"
)
```
