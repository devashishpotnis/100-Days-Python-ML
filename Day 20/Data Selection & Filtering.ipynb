```python
import pandas as pd
```

```python
#Create Sample DataFrame
data = {
    "Name": ["Devashish", "Omkar", "Anand", "Prathmesh", "Shree"],
    "Age": [23, 28, 21, 30, 26],
    "City": ["Mumbai", "Pune", "Delhi", "Mumbai", "Nashik"],
    "Salary": [30000, 45000, 25000, 50000, 40000]
}

df = pd.DataFrame(data)

print("Full Data:\n", df)
```

```python
#head(), tail(), sample()

print("\nFirst 3 rows:\n", df.head(3))

print("\nLast 2 rows:\n", df.tail(2))

print("\nRandom Sample:\n", df.sample(2))
```
