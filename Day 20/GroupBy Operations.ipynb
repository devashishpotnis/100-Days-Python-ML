```python
import pandas as pd
```

```python
data = {
    "Store": ["A", "B", "A", "B", "C", "A", "C"],
    "Item": ["Milk", "Bread", "Butter", "Milk", "Bread", "Bread", "Butter"],
    "Sales": [200, 150, 300, 250, 100, 180, 220],
    "Quantity": [10, 8, 15, 12, 5, 9, 11]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Group by Store
group = df.groupby("Store")
print(group)
```

```python
#Aggregations
#sum()
print(df.groupby("Store")["Sales"].sum())
```

```python
#mean()
print(df.groupby("Store")["Sales"].mean())
```

```python
#count()
print(df.groupby("Store")["Sales"].count())
```

```python
#max() and min()
print(df.groupby("Store")["Sales"].max())
print(df.groupby("Store")["Sales"].min())
```
