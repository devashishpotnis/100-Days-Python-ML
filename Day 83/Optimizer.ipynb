```python
import torch 
import torch.nn as nn
import torch.optim as optim
```

```python
# Simple model
model = nn.Linear(1, 1)
```

```python
# Optimizer (Adam)
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
# Dummy data
x = torch.tensor([[1.0]])
y = torch.tensor([[2.0]])
```

```python
losses = []   # store loss values

# Training
for i in range(20):
    optimizer.zero_grad()

    output = model(x)
    loss = (output - y)**2

    loss.backward()
    optimizer.step()

    losses.append(loss.item())
    print("Loss:", loss.item())
```

```python
import matplotlib.pyplot as plt
# 🔹 Visualization
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Decreasing Graph")
plt.show()
```
