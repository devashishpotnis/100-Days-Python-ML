```python
import torch.nn as nn
import torch.nn as nn
import torch.nn.init as init
```

```python
#Default Initialization (PyTorch gives random)

layer = nn.Linear(2, 2)

print("Default Weights:")
print(layer.weight)
```

```python
#Xavier Initialization (Best for Sigmoid/Tanh)

layer = nn.Linear(2, 2)

init.xavier_uniform_(layer.weight)

print("Xavier Weights:")
print(layer.weight)
```
