```python
import matplotlib.pyplot as plt
```

```python
#Box Plot
data = [10,20,30,40,50,100]

plt.boxplot(data)
plt.title("Box Plot Example")
plt.show()
```

```python
#Violin Plot
plt.violinplot(data)
plt.title("Violin Plot")
plt.show()
```

```python
#Stack Plot
days = [1,2,3,4]
a = [3,4,5,6]
b = [2,2,3,3]

plt.stackplot(days, a, b, labels=["Product A","Product B"])
plt.legend()
plt.show()
```

```python
#Area Chart
plt.fill_between(days, a)
plt.title("Area Chart")
plt.show()
```
