```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,25,30]

plt.plot(x, y, label="Sales")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Plot 1")

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Plot 2")

plt.show()

```

```python
fig, ax = plt.subplots(1,2, figsize=(8,5))

ax[0].plot(x,y)
ax[0].set_title("Line Plot")

ax[1].bar(x,y)
ax[1].set_title("Bar Chart")

plt.show()


```
