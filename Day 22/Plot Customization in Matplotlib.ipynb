```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,25,30]

plt.plot(x, y, label="Sales")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.plot(
    x, y,
    color="red",
    linewidth=3,
    marker="o",
    markersize=10,
    label="Sales"
)

plt.legend()
plt.show()
```
