```python
import math
```

```python
# Inputs
x1, x2 = 2.0, 3.0
y_true = 1
```

```python
# Weights
w1, w2 = 0.5, -0.2
w3, w4 = 0.8, 0.4
w5, w6 = 1.0, -1.5

b1, b2, b3 = 0.1, -0.1, 0.2

lr = 0.1  # learning rate
```

```python
# Functions
def relu(z):
    return max(0, z)

def sigmoid(z):
    return 1 / (1 + math.exp(-z))
```

```python
# FORWARD
h1 = relu(x1*w1 + x2*w2 + b1)
h2 = relu(x1*w3 + x2*w4 + b2)

z = h1*w5 + h2*w6 + b3
y_pred = sigmoid(z)

loss = -(y_true*math.log(y_pred) + (1-y_true)*math.log(1-y_pred))

print("Before:", y_pred, "Loss:", loss)
```

```python
# Output error
error = y_pred - y_true   # very important shortcut
```

```python
# Update output weights
w5 -= lr * error * h1
w6 -= lr * error * h2
b3 -= lr * error
```

```python
# Update hidden weights (approx)
w1 -= lr * error * x1
w2 -= lr * error * x2
w3 -= lr * error * x1
w4 -= lr * error * x2
b1 -= lr * error
b2 -= lr * error
```

```python
# FORWARD AGAIN
h1 = relu(x1*w1 + x2*w2 + b1)
h2 = relu(x1*w3 + x2*w4 + b2)

z = h1*w5 + h2*w6 + b3
y_pred = sigmoid(z)

loss = -(y_true*math.log(y_pred) + (1-y_true)*math.log(1-y_pred))

print("After :", y_pred, "Loss:", loss)
```
