```python
import math
```

```python
# Input Layer
x1, x2 = 2.0, 3.0
```

```python
# Hidden Layer Weights & Bias
w11, w12, b1 = 0.5, -0.2, 0.1
w21, w22, b2 = 0.8, 0.4, -0.1
```

```python
# ReLU Function
def relu(z):
    return max(0, z)
```

```python
# Hidden Layer Computation
z1 = (x1 * w11 + x2 * w12) + b1
h1 = relu(z1)

z2 = (x1 * w21 + x2 * w22) + b2
h2 = relu(z2)
```

```python
# Output Layer Weights & Bias
w_out1, w_out2, b_out = 1.0, -1.5, 0.2
```

```python
# Output Computation
z_out = (h1 * w_out1 + h2 * w_out2) + b_out
output = 1 / (1 + math.exp(-z_out))  # Sigmoid
```

```python
# Final Output
print("Output:", output)
```

```python
# Prediction
if output >= 0.5:
    print("Prediction: Positive Class")
else:
    print("Prediction: Negative Class")
```

```python
import matplotlib.pyplot as plt
plt.figure()

# Neuron positions
input_y = [1, 2]
hidden_y = [1.5, 1.5]
output_y = [1.25]

# Plot neurons
for y in input_y:
    plt.scatter(0, y)

for y in hidden_y:
    plt.scatter(1, y)

for y in output_y:
    plt.scatter(2, y)

# Connections
for y1 in input_y:
    for y2 in hidden_y:
        plt.plot([0, 1], [y1, y2])

for y1 in hidden_y:
    for y2 in output_y:
        plt.plot([1, 2], [y1, y2])

plt.title("Feed Forward Neural Network")
plt.axis('off')
plt.show()
```

```python
values = [x1, x2, h1, h2, output]
labels = ["x1", "x2", "h1", "h2", "Output"]

plt.figure()
plt.plot(values, marker='o')
plt.xticks(range(len(labels)), labels)
plt.title("Feed Forward Value Flow")
plt.xlabel("Neurons")
plt.ylabel("Values")
plt.grid()
plt.show()
```
